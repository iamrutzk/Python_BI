```python
from Bio.Seq import Seq
print("Original DNA seq:",bio_seq)
#Transcribe DNA to RNA
rna_seq=bio_seq.transcribe()
print("RNA seq:",rna_seq)
## translate
prt_seq=rna_seq.translate()
print("Protein seq:",prt_seq)

# complement
cDNA=bio_seq.complement()
print("cDNA seq:",cDNA)

# reverse
rev_compl=bio_seq.reverse_complement()
print("rev seq:",rev_compl)
```

```python
from Bio import pairwise2

# Perform pairwise sequence alignment
seq1 = Seq("AGTACACTGG")
seq2 = Seq("AGTACGTGAC")

alignments = pairwise2.align.globalxx(seq1, seq2, one_alignment_only=True)

for alignment in alignments:
    print("Aligned Sequence 1:", alignment.seqA)
    print("Aligned Sequence 2:", alignment.seqB)
```

```python
from Bio import Align
from Bio.Seq import Seq

# Create a pairwise aligner
aligner = Align.PairwiseAligner()
option:aligner.mode = 'local'
# Define two sequences
seq1 = Seq("GATTACA")
seq2 = Seq("GCATGCU")

# Perform the alignment
alignments = aligner.align(seq1, seq2)

# Print the alignment and score
for alignment in alignments:
    print(alignment)
    print(f"Score: {alignment.score}")
```
