```python
# R,W,A
#R+W------> r+
with open("/content/demo.fasta","r+") as file:
  content=file.read()
  print("Initial Content:",content)

  file.write( "\n This line is added with r+ mode")

  file.seek(0)
  content=file.read()
  print("Updated Content:",content)
```

```python
# W+
#write and read
with open('output.txt', 'w+') as file:
    file.write("This is the first line written with w+ mode.\n")
    file.seek(0)
    content = file.read()
    print("Content after writing:", content)


    file.write("This is the second line written with w+ mode.\n")
```

```python
with open('output.txt', 'a+') as file:
    file.write("This is the first line written with a+ mode.\n")
    file.seek(0)
    content = file.read()
    print("Content after appending:", content)


    file.write("This is the second line written with a+ mode.\n")
```

```python
'''
Assignmnet:
Step 1: Open one.txt in write mode and write some content to it, then close the file.
Step 2 & 3: Open one.txt in read mode, read its contents, and then write those contents to two.txt.
Step 4: Use the os.remove() function to delete one.txt from the filesystem.
'''

import os

# Step 1: Open one.txt in write mode, write content, and close the file
with open("one.txt", "w") as file_one:
    file_one.write("This is the content of one.txt. We are practicing file handling in Python.\n")
    file_one.write("File handling is useful in bioinformatics and other fields.\n")

print("Step 1: Content written to one.txt")

# Step 2: Open one.txt in read mode, read content, and write into two.txt
with open("one.txt", "r") as file_one:
    content = file_one.read()

with open("two.txt", "w") as file_two:
    file_two.write(content)

print("Step 2: Content from one.txt copied to two.txt")

# Step 3: Delete one.txt using os.remove
if os.path.exists("one.txt"):
    os.remove("one.txt")
    print("Step 3: one.txt deleted")
else:
    print("one.txt does not exist!")
```
