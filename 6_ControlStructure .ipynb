```python
# Example: Printing amino acid positions in a sequence
sequence = "MSYVVAIPDMLASAAADVAGIGSSVG"  # Example protein sequence

# Start with position 1
i = 1

# Loop through until the end of the sequence
while i <= len(sequence):
    print(f"Position {i}: {sequence[i-1]}")
    i += 1
```

```python
#Assignmnet: Processing DNA Sequences: Removing Occurrences of 'CG' with a While Loop"
# Simulating a DNA sequence
seq = "ATCGATCGATCG"
processed_sequence = ""

# Define a while loop to process the DNA sequence
while "CG" in seq:
    # Find the first occurrence of "CG" in the sequence
    index = seq.find("CG")

    # Extract the substring before the "CG" and add it to the processed sequence
    processed_sequence += seq[:index]

    # Update the DNA sequence to exclude the processed part
    seq = seq[index + 2:]

# Add the remaining part of the DNA sequence to the processed sequence
processed_sequence += seq

# Print the original and processed DNA sequences
print("Original DNA sequence:", "ATCGATCGATCG")
print("Processed DNA sequence:", processed_sequence)


```
