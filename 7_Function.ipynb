```python
from shutil import register_unpack_format
# Total seq len
def total_seq_len(*args):
  total=0
  for lenght in args:
    total+=lenght
  return f"Total Sequence Length:{total} bp"
```

```python
def GC_multiple_seq(*sequence):
  for seq in sequence:
    gc_count=seq.count('G')+seq.count('C')
    gc_content=gc_count/len(seq)
    print(f"Sequence:{seq},  GC_Content:{gc_content}")


GC_multiple_seq("ACCAGATAGCAAG","AGTAAGCA","ACGATAGAAGCAAGACAGACAGAAGTAG")
```
